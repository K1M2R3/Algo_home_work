### 2. Исследовать время поиска различными способами и сделать вывод об алгоритмической сложности. Проверить на массивах различного объема (10, 10_000, 10_000_000 элементов).

```python
def simple(arr, target):
    for i in range(len(arr)):
        if target == arr[i]:
            return i
    return False

def binarium(arr, target):
    val = arr.sort()
    start = 0
    end = len(arr)-1
    index = -1
    while start <= end and index == -1:
        mid = (start+end)//2
        if arr[mid]==target:
            index = mid
        else:
            if arr[mid]>target:
                end = mid - 1
            else:
                start = mid + 1
    return index

def bubbleSort(arr):
    print(arr)
    if len(arr) <=1:
        return arr
    for i in range(len(arr)):
        for j in range(0, len(arr)-i-1):
            if arr[j]>arr[j+1]:
                arr[j], arr[j+1]=arr[j+1], arr[j]
    return arr
```

```python
import random 
list = [random.randint(0,10) for _ in range(10)]
print(list)
simple(list, 10)
```

```python
list = [random.randint(0,10) for _ in range(10)]
print(bubbleSort(list))
binarium(list, 5)
```

```python
import random as rand
import timeit

no_sort_10 = [rand.randint(0, 100) for _ in range(10)]
no_sort_10K = [rand.randint(0, 100) for _ in range(10**4)]
no_sort_1M = [rand.randint(0, 100) for _ in range(10**6)]
no_sort = [no_sort_10,no_sort_10K,no_sort_1M]
print(simple(no_sort[2], 20))
time =[]
for i in range(len(no_sort)):
    execution_time = timeit.timeit(lambda: simple(no_sort[i], 20), number=1000)
    time.append(execution_time)
    print(f"Среднее время: {execution_time:.6f} милисекунд")
```

```python
import matplotlib.pyplot as plt
X = [10, 10**4, 10**6]
plt.plot(X, time)
plt.show()
```

#### Судя по графику выше, можно судить о том что линейный поиск иммеет сложность O(nlog(n))

```python
time1 = []
for i in range(len(no_sort)):
    execution_time = timeit.timeit(lambda: binarium(no_sort[i], 20), number=1000)
    time1.append(execution_time)
    print(f"Среднее время: {execution_time:.6f} милисекунд")
```
