Класс **CustomList** — упрощённый аналог списка с дополнительными методами

```python
class CustomList:
    def __init__(self):
        self.data = []

    def append(self, item):
        """Добавляет элемент в конец списка."""
        self.data.append(item)

    def remove(self, item):
        """Удаляет первое вхождение элемента."""
        if item in self.data:
            self.data.remove(item)
        else:
            raise ValueError(f"Элемент {item} не найден")

    def reverse(self):
        """Разворачивает список."""
        self.data = self.data[::-1]

    def __str__(self):
        return str(self.data)

# Пример использования
cl = CustomList()
cl.append(10)
cl.append(20)
cl.reverse()
print(cl)  # [20, 10]
```

класс **CaseInsensitiveDict**, где ключи нечувствительны к регистру

```python
class CaseInsensitiveDict:
    def __init__(self):
        self.data = {}

    def __setitem__(self, key, value):
        """Сохраняет ключ в нижнем регистре."""
        self.data[key.lower()] = value

    def __getitem__(self, key):
        """Возвращает значение по ключу без учёта регистра."""
        return self.data[key.lower()]

    def swap_key_value(self):
        """Меняет ключи и значения местами."""
        self.data = {v: k for k, v in self.data.items()}

    def __str__(self):
        return str(self.data)

# Пример использования
cid = CaseInsensitiveDict()
cid["Name"] = "Alice"
print(cid["name"])  # "Alice"
#cid.swap_key_value()
#print(cid)  # {'Alice': 'name'}
```

```python
cid.swap_key_value()
print(cid)  # {'Alice': 'name'}
```

### 2.1 Реализация «Умного списка»
Создайте класс **SmartList**, который: <br> Автоматически удаляет дубликаты при добавлении (append). <br> Имеет метод get_stats(), возвращающий: <br> - количество уникальных элементов, <br> - самый частый элемент.
### *2.2 Множество с историей
Создайте класс **HistorySet**, который: <br> Запоминает все добавленные и удалённые элементы. <br> Имеет метод undo(), отменяющий последнее действие.

```python
class SmartList:
    def __init__(self):
        self.data = [1,1,1,1,1,1234,412,4,65,562,456,2,456,7,6,74,6,5,65,5,55,4,4,44,4,4,3,24,2,3,3]

    def append(self, item):
        if item not in self.data:
            self.data.append(item)
        else:
            print(f'AHA, I find a duplicate! Please, try again!)')

    def remove(self, item):
        if item in self.data:
            self.data.remove(item)
        else:
            print(f"Wrong! This item: {item}, not in list!")
    def get_stats(self):
        print(f'Чилсло уникальных символов в списке: {len(set(self.data))}')
        from collections import Counter
        counter = Counter(self.data)
        most_common = counter.most_common(1)[0][0]
        print(f'Самый уникальный элемент: {most_common}')
    def __str__(self):
        return str(self.data)
    
```

```python
import random
a = SmartList()
for i in range(10):
    a.append(random.randint(0,5))

print(a.data)            # [1, 2, 3, 4, 5] — дубликаты автоматически убраны
print(a.get_stats())
```

```python
class HistoryList:
    def __init__(self):
        self.data = []
        self.history = []
        self.actions = []
    def add(self, item):
        self.data.append(item)
        self.history.append(item)
        self.actions.append("add")
    def delete(self, item):
        if item in self.data:
            self.data.remove(item)
            self.actions.append("delete")
        else: print(f'This {item} is not in {self.data}')
    def undo(self):
        if not self.history:
            print(f'Nothing to return!')
            return
        
        a = self.history.pop()
        b = self.actions.pop()
        if b == 'add':
            self.data.remove(a)
        elif b == 'delete':
            self.data.append(a)
    def __str__(self):
        return str(self.data)
```

```python
d = HistoryList()
d.add(1)
d.add(12)
d.add(999)
d.delete(999)
print(d)
d.undo()
print(d)
d.undo()
print(d)
d.undo()
print(d)
```
