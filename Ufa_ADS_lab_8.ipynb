```python
# Задача о расписании
events = [(1, 4), (3, 5), (0, 6), (5, 7), (3, 9), (5, 8), (4, 10), (6, 8)]
events.sort(key= lambda x: x[1])

selected = []
last_end = 0
for start, end in events:
    if start >= last_end:
        selected.append((start, end))
        last_end = end

print("Выбрали: ", selected)
```

```python
# Задача о рюкзаке
capacity = 10
items = [(3.5, 4), (4, 4.5), (5, 8)]
bag = 0

for price, weight in items:
    if capacity >= weight:
        bag += price
        capacity -= weight

print(f'Награбленного на {bag} денег, свободного места в рюкзаке {capacity}')
```

```python
# Задача о покрытии множества

regions = set(['Московская', "Ленинградская", "Новосибирская", 
           "Свердловская", "Челябинская", "Краснодарский", "Татарстан", "Башкортостан"])

stations = {
    "st1": set(['Московская', "Челябинская", "Татарстан"]),
    "st2": set(['Московская', "Ленинградская", "Новосибирская"]),
    "st3": set(["Свердловская", "Челябинская", "Краснодарский"]),
    "st4": set(["Краснодарский", "Башкортостан"]),
    "st5": set(["Новосибирская", "Башкортостан"])
}

chosen = set()
while regions:
    best_st = None
    reg_covered = set()
    for station, region_st in stations.items():
        covered = regions&region_st
        if len(covered) > len(reg_covered):
            best_st = station
            reg_covered = covered
    if best_st:
        chosen.add(best_st)
        regions -= reg_covered
print("Лучший набор станций:", chosen)
```

```python
# Задача размена монетами
def coin_change(n: int, coins: list) -> list:
    coins.sort(reverse=True)
    result = []
    for coin in coins:
        while n>=coin:
            result.append(coin)
            n -= coin
    return result

coin_change(9, [10,5,2,1])
```

## Задача 3. Контрпример для жадного подхода

Придумайте пример, где жадный алгоритм не даёт оптимального решения. Например, задача о размене:

Монеты: [1, 3, 4], сумма: 6
Жадный подход даст: 4 + 1 + 1 = 3 монеты
Оптимум: 3 + 3 = 2 монеты

Смоделируйте такой пример программно и проверьте результат жадного и оптимального алгоритма.



```python
import itertools

cost = [
  [9, 2, 7],
  [6, 4, 3],
  [5, 8, 1]
]


def full_iteration(costs):
    n = len(costs)
    min_cost = float('inf')
    best_assignment = []
    for perm in itertools.permutations(range(n)):
        total = sum(costs[i][perm[i]] for i in range(n))
        if total < min_cost:
            min_cost = total
            best_assignment = perm
    return min_cost, best_assignment

min_cost, person = full_iteration(cost)
print(f"Минимальная стоимость: {min_cost}, Назначения: {person}")
```

#### Жадный алгоритм, вероятнее всего выдал бы значение 10 с назнаением (1,2,0). Поэтому полный перебор, более выгодный метод.

```python
def lines(m, segments):
    segments.sort()
    current_end = 0
    result = []
    i = 0
    n = len(segments)
    
    while current_end < m and i < n:
        next_end = current_end
        best_index = -1
        while i < n and segments[i][0] <= current_end:
            if segments[i][1] > next_end:
                next_end = segments[i][1]
                best_index = i
            i += 1
        if best_index == -1:
            return [] 
        result.append(segments[best_index])
        current_end = next_end
    
    return result if current_end >= m else []
```

```python
m = 10
segments = [(0, 5), (4, 7), (6, 10), (1, 3)]
n = lines(m, segments)
print(n)
```

```python
from collections import deque

def coin_change_greed(n: int, coins: list) -> list:
    coins.sort(reverse=True)
    result = []
    for coin in coins:
        while n>=coin:
            result.append(coin)
            n -= coin
    return result


def bfs_coin_change(coins, target):
    visited = set()
    queue = deque([(0, 0)])  # (current_sum, steps)
    while queue:
        current_sum, steps = queue.popleft()
        if current_sum == target:
            return steps
        if current_sum > target or current_sum in visited:
            continue
        visited.add(current_sum)
        for coin in coins:
            next_sum = current_sum + coin
            queue.append((next_sum, steps + 1))
    return -1
```

```python
coins = [1,3,4]
target = 6
greed = coin_change_greed(target, coins)
no_greed = bfs_coin_change(coins, target)
print(f'Жадный алгоритм:{len(greed)}')
print(f'В ширину:{no_greed}')
```
