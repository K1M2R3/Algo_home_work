```python
# Числа Фибоначчи

def fibonacci(n):
    dp = [0] * (n+1)
    if n > 0:
        dp[1] = 1
    for i in range(2, n+1):
        dp[i] = dp[i-1] + dp[i-2]
    return dp

print(fibonacci(100))  
```

```python
# Размен монет

def min_coins(n, coins):
    dp = [float("inf")] * (n + 1)
    dp[0] = 0

    for coin in coins:
        for i in range(coin, n+1):
            dp[i] = min(dp[i], dp[i-coin]+1)
    
    return dp[n] if dp[n] != float('inf') else -1

print(min_coins(6, [1,3,4]))
```

```python
# Задача о рюкзаке

def bag(weight, prices, capacity):
    n = len(weight)
    dp = [[0]*(capacity+1) for _ in range(n+1)]

    for i in range(1, n+1):
        for w in range(1, capacity+1):
            if weight[i - 1] <= w:
                dp[i][w] = max(dp[i-1][w], dp[i-1][w-weight[i-1]]+prices[i-1])
            else:
                dp[i][w] = dp[i-1][w]
    return dp[n][capacity]

print(bag([1, 4, 5, 8], [6, 3, 4, 5], 10))
```

## Задание 3. Мемоизация

Добавить в задачи о числах Фибоначчи и размене монет мемоизацию

```python
def letter_swap(s1, s2):
    if len(s1) < len(s2):
        return letter_swap(s2, s1)
    
    previous = list(range(len(s2) + 1))
    
    for i, c1 in enumerate(s1, 1):
        current = [i] * (len(s2) + 1)
        for j, c2 in enumerate(s2, 1):
            cost = 0 if c1 == c2 else 1
            current[j] = min(
                previous[j] + 1,     # Удаление
                current[j-1] + 1,    # Вставка
                previous[j-1] + cost # Замена
            )
        previous = current
    
    return previous[-1]
```

```python
start = 'кот'
end = 'скит'
print(letter_swap(start, end))
```

```python
def max_Subarray_Sum(my_array, array_size): 

    maxTillNow = my_array[0] 
    maxEnding = 0 

    for n in range(0, array_size): 
        maxEnding = maxEnding + my_array[n] 

        if maxEnding < 0: 
            maxEnding = 0 
         
        elif(maxTillNow < maxEnding): 
            maxTillNow = maxEnding 
             
    return maxTillNow 

my_array = [-2, -3, 4, -1, -2, 5, -3] 

print("Maximum Subarray Sum:", max_Subarray_Sum(my_array, len(my_array))) 
```

```python
from functools import lru_cache

@lru_cache(maxsize=None)
def fib_memo(n):
    if n <= 1:
        return n
    return fib_memo(n-1) + fib_memo(n-2)


print(fib_memo(100))
```

```python
def min_coins_memo(coins, amount):
    @lru_cache(maxsize=None)
    def helper(remaining):
        if remaining == 0:
            return 0
        if remaining < 0:
            return float('inf')
        return 1 + min(helper(remaining - coin) for coin in coins)
    
    result = helper(amount)
    return result if result != float('inf') else -1

coins = [1, 3, 4]
print(min_coins_memo(coins, 6))
```
